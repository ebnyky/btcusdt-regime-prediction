```python
from pathlib import Path
from zipfile import ZipFile
import hashlib

archive = Path(
    "/kaggle/working/BTCUSDT_k3_90to30_stride1_dataset_v1_0.zip"
)

digest = hashlib.sha256()

with archive.open("rb") as handle:
    for chunk in iter(lambda: handle.read(1024 * 1024), b""):
        digest.update(chunk)

with ZipFile(archive) as z:
    names = z.namelist()
    corrupt_file = z.testzip()

print("Archive exists:", archive.is_file())
print("Archive size bytes:", archive.stat().st_size)
print("Archive SHA-256:", digest.hexdigest())
print("Total archived files:", len(names))
print("JPEG images:", sum(name.lower().endswith((".jpg", ".jpeg")) for name in names))
print("CSV manifests:", sum(name.lower().endswith(".csv") for name in names))
print("JSON files:", sum(name.lower().endswith(".json") for name in names))
print("Corrupt file:", corrupt_file)
```
